# H1

```python
import cv2
import mujoco


env = gym.make(
        "h1-reach-v0",
        render_mode="rgb_array",
)
env.reset()
print(env.observation_space)
data = env.unwrapped.named.data
print(data.qpos)
img = env.render()
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
import matplotlib.pyplot as plt
plt.imshow(rgb_img)
plt.show()

env.data.qpos[3:7] = [0.7071, 0, 0.7071, 0]
mujoco.mj_forward(env.model, env.data)

img = env.render()
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb_img)
plt.show()

print(data.qpos)
print(data.qvel)
```
