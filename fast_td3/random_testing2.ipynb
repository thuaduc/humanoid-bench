```python
from fast_td3.actors.gnn.egnn import EGNN
import torch

egnn = EGNN(
    in_node_nf = 2,
	hidden_nf = 64,
	out_node_nf = 1,
	in_edge_nf = 0,
	device = "cuda",
	batch_size = 32,
	act_fn = torch.nn.ReLU(),
	n_layers = 4,
	robot = "h1",
	env_name="h1_run_v0",
)

index = egnn.generate_index(1)
print(len(index[0]))

# test if all edges are bidirectional
edge_set = set()
for src, dst in zip(index[0], index[1]):
	edge_set.add((src.item(), dst.item()))

# Check which edges are missing their reverse
missing_edges = []
for src, dst in zip(index[0], index[1]):
	if (dst.item(), src.item()) not in edge_set:
		missing_edges.append((src.item(), dst.item()))

if missing_edges:
	print(f"Missing reverse edges for: {missing_edges}")
else:
	print("All edges are bidirectional")
```
